```python
#Data cleaning
#data type conversion
events["super_tackle"].value_counts()
```

```python
events["super_tackle"]=events['super_tackle'].astype(int)
```

```python
events["super_tackle"]=pd.to_numeric(events["super_tackle"],errors="coerce")
```

```python
events["super_raid"]=pd.to_numeric(events["super_raid"])
```

```python
#data type conversion
#categorical data
#select all the object columns
categorical_columns=events.select_dtypes(include=["object"]).columns
```

```python
categorical_columns
```

```python
#convert into category
for c in categorical_columns:
    events[c]=events[c].astype("category")
```

```python
#dropna()
events.dropna(axis=1,how="all",inplace=True)
```

```python
events.columns
```

```python
events.shape
```

```python
#Handling the missing values
```

```python
events.isnull().sum()
```

```python
#select all the columns that ends with points
# x="hello"
# print(x.endswith("p"))
```

```python
points_column=[col for col in events.columns if col.endswith("points")]
```

```python
points_column
```

```python
p=[]
for col in events.columns:
    if col.endswith("points"):
       p.append(col)
print(p)
```

```python
events[points_column]=events[points_column].fillna(0)
```

```python
events.isnull().sum()
```

```python
#ends with id
id_column=[col for col in events.columns if col.endswith("_id") or col.endswith(".id")]
```

```python
id_column
```

```python
#replace -
events[id_column]=events[id_column].fillna("-")
```

```python
events[id_column].isnull().sum()
```

```python
events.info()
```

```python
#numerical data
#null--->median
numerical_columns=events.select_dtypes(include=[np.number]).columns
```

```python
numerical_columns
```

```python
for column in numerical_columns:
    events[column].fillna(events[column].median(),inplace=True)
```

```python
events[numerical_columns].isnull().sum()
```

```python
events.info()
```

```python
categorical_columns=events.select_dtypes(include=["category"]).columns
```

```python
categorical_columns
```

```python
for column in categorical_columns:
    events[column].fillna(events[column].mode()[0],inplace=True)
```

```python
events.isnull().sum()
```

```python
#ensure the data consistency in numerical columns
#points--->-ve
p=["defending_points","raid_points"]
```

```python
for c in p:
    events[c]=events[c].apply(lambda x:max(x,0))
```
